28. Perform matrix multiplication on two compatible matrices.

```python
# In order to perform matrix maltiplication the columns of first matrix should be the same size as the row of second matrix
matrix_A = np.arange(1, 7).reshape(2, 3)    # (row=2, column=3)
matrix_B = np.arange(1, 13).reshape(3, 4)   # (row=3, column=4)
matrix_C = np.matmul(matrix_A, matrix_B)
```

```python
print(matrix_A)
print(matrix_A.shape)
print(matrix_B)
print(matrix_B.shape)
print(matrix_C)
print(matrix_C.shape)
```

30. Solve the system of linear equations: 2x + y = 5, x - y = 1.

```python
# Coefficient Matrix - A
A = np.array([[2, 1],
              [1, -1]])

# Constants Vector - b
b = np.array([5, 1])

# Variable Vector - x
x = np.linalg.solve(A, b)
x
```

```python
# Validation = Ax=b
np.linalg.matmul(A, x)
```
