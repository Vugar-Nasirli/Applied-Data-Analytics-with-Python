**6.8** Monte Carlo: Estimate π by randomly sampling 100,000 points in the unit square and counting how many fall inside the unit quarter‑circle.

```python
# NOTE: Gemini Generated Answer
# Set a random seed for reproducible results
np.random.seed(42)

num_points = 100000

# 1. Generate random x and y coordinates uniformly distributed between 0 and 1
x = np.random.rand(num_points)
y = np.random.rand(num_points)

# 2. Calculate the squared distance from the origin (0,0) for all points
# Squaring avoids calculating a costly square root (sqrt) operation
distance_squared = x**2 + y**2

# 3. Count how many points have a distance <= 1
points_inside = np.sum(distance_squared <= 1)

# 4. Estimate pi
pi_estimate = 4 * (points_inside / num_points)

print(f"Points Inside: {points_inside:,} out of {num_points:,}")
print(f"Estimated π:   {pi_estimate}")
print(f"Actual π:      {np.pi}")
print(f"Absolute Error: {abs(np.pi - pi_estimate):.6f}")
```

**9.1 Finance**: Prices of a stock for 10 days are `p = [100, 102, 101, 103, 105, 104, 106, 107, 110, 108]`. Compute **daily returns** (percentage) and **cumulative return**.

```python
prices = np.array([100, 102, 101, 103, 105, 104, 106, 107, 110, 108])
windows = np.lib.stride_tricks.sliding_window_view(x=prices, window_shape=2)
yesterdays = windows[:, 0]
todays = windows[:, 1]
daily_returns = ((todays - yesterdays) / yesterdays) * 100
print("Daily Returns:\n", daily_returns)
print("Cumulative Returns:\n", np.cumsum(daily_returns))
```

**9.4 A/B Testing**: Variant A converts 48/500 users; Variant B converts 62/500. Compute conversion rates and the **absolute lift** (B−A). Simulate 10,000 bootstrap samples of the lift.

```python
# NOTE: Gemini generated answer.
# Experimental Data
n_A, conv_A = 500, 48
n_B, conv_B = 500, 62

# Create raw user outcome arrays (1 = converted, 0 = did not convert)
data_A = np.zeros(n_A)
data_A[:conv_A] = 1

data_B = np.zeros(n_B)
data_B[:conv_B] = 1

# Setup bootstrapping parameters
np.random.seed(42)
n_bootstraps = 10000

# Draw 10,000 bootstrap samples for both variants (with replacement)
# Generating a 10000x500 matrix and calculating the mean along rows (axis=1)
boot_cr_A = np.random.choice(data_A, size=(n_bootstraps, n_A), replace=True).mean(axis=1)
boot_cr_B = np.random.choice(data_B, size=(n_bootstraps, n_B), replace=True).mean(axis=1)

# Calculate simulated absolute lifts
boot_lifts = boot_cr_B - boot_cr_A

# Compute 95% Confidence Interval (2.5th to 97.5th percentile)
ci_lower, ci_upper = np.percentile(boot_lifts, [2.5, 97.5])

# Compute probability that Variant B is truly better than Variant A
prob_b_better = np.mean(boot_lifts > 0)

print(f"Observed Absolute Lift: {conv_B/n_B - conv_A/n_A:+.4%}")
print(f"Mean Bootstrap Lift:   {np.mean(boot_lifts):+.4%}")
print(f"95% Confidence Interval: [{ci_lower:+.4%}, {ci_upper:+.4%}]")
print(f"Probability that B > A:  {prob_b_better:.2%}")
```
