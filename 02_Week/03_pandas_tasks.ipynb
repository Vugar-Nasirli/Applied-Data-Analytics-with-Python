### Bölmə 3: Məlumatın Seçilməsi, İndeksləmə və Filtrlənməsi (Data Selection & Filtering)

```python
# 12. Yalnız 'Name' (Ad) sütununu tək bir Series obyekti kimi seçin və ilk 5 dəyəri göstərin.
df["Name"].head()
```

```python
# 13. 'Name' və 'Age' (Yaş) sütunlarını birlikdə, DataFrame formatında seçin.
df[["Name", "Age"]]
```

```python
# 14. Yaşı 30-dan böyük olan sərnişinləri filtrləyərək yeni bir DataFrame yaradın.
mask = df["Age"] > 30
df[mask]
```

```python
# 15. Məntiqi operatorlardan (&) istifadə edərək yaşı 30-dan böyük olan VƏ qadın ('female') sərnişinləri filtrləyin.
mask = (df["Age"] > 30) & (df["Sex"] == "female")
df[mask]
```

```python
# 16. Məntiqi operatorlardan (|) istifadə edərək yaşı 18-dən kiçik olan VƏ YA 1-ci sinif (Pclass == 1) səyahət edənləri tapın.
mask = (df["Age"] < 18) | (df["Pclass"] == 1)
df[mask]
```

```python
# 17. `loc` metodundan istifadə edərək indeks nömrəsi 10-dan 20-yə qədər (20 daxil) olan sətirləri və yalnız 'Name', 'Age' sütunlarını seçin.
df.loc[10:20, ["Name", "Age"]]
```

```python
# 18. `iloc` metodundan istifadə edərək ilk 5 sətri və ilk 3 sütunu rəqəmsal mövqelərinə görə seçin.
df.iloc[:5, :3]
```

```python
# 19. Sərnişinlərin bilet qiyməti (Fare) 100-dən çox olanların ümumi sayını tapın.
mask = df["Fare"] > 100
len(df[mask])
```

```python
# 20. 'Embarked' (Liman) sütunu 'C' və ya 'S' olan sətirləri `.isin()` metodu ilə filtrləyin.
mask = df["Embarked"].isin(["C", "S"])
df[mask]
```

### Bölmə 4: Məlumatın Təmizlənməsi və Transformasiyası (Data Cleaning & Transformation)

```python
# 21. Hər bir sütunda neçə ədəd çatışmayan (NaN) dəyər olduğunu hesablayın.
df.isna().sum(axis=0)
```

```python
# 22. İçərisində heç olmasa bir dənə çatışmayan dəyər olan bütün sətirləri (dropna) müvəqqəti silin və qalan sətir sayını göstərin.
df.dropna(axis=0, how="any", inplace=False).shape[0]
```

```python
# 23. 'Cabin' sütununda 70%-dən çox məlumat çatışmadığı üçün bu sütunu df-dən tamamilə silin (drop).
df["Cabin"].value_counts(normalize=True, dropna=False)[:5]
```

```python
df.drop(columns="Cabin", inplace=True)
df
```

```python
# 24. 'Age' sütunundakı boş dəyərləri (NaN) bütün sərnişinlərin median yaşı ilə əvəzləyin (fillna).
df["Age"].value_counts(dropna=False)
```

```python
median_age = df["Age"].median()
df.fillna({"Age":median_age}, inplace=True)
df["Age"].value_counts(dropna=False)
```

```python
# 25. 'Embarked' sütunundakı boş dəyərləri ən çox təkrar olunan dəyər (mode) ilə doldurun.
df["Embarked"].value_counts(dropna=False)
```

```python
mode_embarked = df["Embarked"].mode()[0]
df.fillna({"Embarked":mode_embarked}, inplace=True)
df["Embarked"].value_counts(dropna=False)
```

```python
# 26. Məlumat dəstində təkrarlanan sətirlərin (duplicates) olub-olmadığını yoxlayın və varsa silin.
df.duplicated().sum()
```

```python
# 27. Sütun adlarında aydınlıq yaratmaq üçün 'Pclass' sütununun adını 'Passenger_Class' olaraq dəyişdirin (rename).
df.rename(columns={"Pclass":"Passenger_Class"}, inplace=True)
df
```

```python
# 28. 'Sex' sütunundakı 'male' və 'female' kateqoriyalarını riyazi analiz üçün müvafiq olaraq 0 və 1 rəqəmləri ilə əvəzləyin (map/replace).
df.replace({"Sex":{"male":0, "female":1}}, inplace=True)
df
```

```python
# 29. Ailə ölçüsünü göstərən yeni bir 'Family_Size' sütunu yaradın ('SibSp' və 'Parch' sütunlarının cəmi + 1).
df["Family_Size"] = df["SibSp"] + df["Parch"] + 1
df
```

```python
# 30. 'Fare' sütununun veri tipini float64-dən int32-yə (tam ədəd) çevirin (astype).
df["Fare"] = df["Fare"].astype(dtype="int32")
df
```

### Bölmə 5: Qruplaşdırma və Aqreqasiya (GroupBy & Aggregation)

```python
# 31. Sərnişin sinfinə ('Passenger_Class') görə qruplaşdırıb (groupby), hər sinfin orta bilet qiymətini tapın.
df.groupby(by="Passenger_Class").aggregate(avg_fare=("Fare", "mean"))
```

```python
# 32. Cinsiyyətə görə sağ qalma ('Survived') dərəcəsini (orta dəyərini) faiz ifadəsi ilə hesablayın.
df.groupby(by="Sex").aggregate({"Survived":"mean"}) * 100
```

```python
# 33. Çoxölçülü qruplaşdırma: Həm 'Passenger_Class', həm də cinsiyyətə görə qruplaşdırıb 'Age' sütununun medianını tapın.
df.groupby(by=["Passenger_Class", "Sex"]).aggregate(median_age=("Age", "median"))
```

```python
# 34. `.agg()` metodundan istifadə edərək 'Fare' sütununun həm minimum, həm maksimum, həm də orta dəyərini tək bir cədvəldə göstərin.
df["Fare"].aggregate(["min", "mean", "median", "max"])
```

```python
# 35. 'Embarked' sütunundakı kateqoriyaların tezliyini (hansı limandan neçə sərnişin minib) hesablayın (value_counts).
df["Embarked"].value_counts()
```

```python
# 36. Qruplaşdırma vasitəsilə ən çox bilet ödənişi edən sərnişin(lər)in hansı sinifdə (Pclass) olduğunu təyin edin.
df.groupby(by="Passenger_Class").aggregate(Total_Fare=("Fare", "sum"), Passenger_Count=("Fare", "count"))
```

### Bölmə 6: Sıralama və İndeks İdarəetməsi (Sorting & Indexing)

```python
# 37. Məlumat dəstini 'Age' sütununa görə artan ardıcıllıqla sıralayın (sort_values).
df.sort_values(by="Age")
```

```python
# 38. Məlumat dəstini əvvəlcə 'Passenger_Class' (artan), daha sonra 'Fare' (azalan) ardıcıllıqla sıralayın.
df.sort_values(by=["Passenger_Class", "Fare"], ascending=[True, False])
```

```python
# 39. DataFrame-in əsas indeksini 'PassengerId' sütunu olaraq təyin edin (set_index).
df.set_index(keys="PassengerId", inplace=True)
df
```

```python
# 40. İndeksi sıfırlayın (reset_index) və köhnə indeksi məlumat kimi df-də saxlayın.
df.reset_index(drop=False, inplace=True)
df
```

### Bölmə 7: Məlumatların Birləşdirilməsi (Merge, Join, Concat)
Bu bölmə üçün xüsusi olaraq iki kiçik DataFrame yaradılmışdır. Aşağıdakı xananı işə salın və istifadə edin.

```python
df1 = pd.DataFrame({'ID': [1, 2, 3], 'Ad': ['Əli', 'Vəli', 'Aysel']})
df2 = pd.DataFrame({'ID': [2, 3, 4], 'Maaş': [500, 600, 700]})
```

```python
# 41. df1 və df2-ni sətir oxu üzrə alt-alta birləşdirin (concat, axis=0) və indeksləri yeniləyin (ignore_index=True).
pd.concat(objs=[df1, df2], axis=0, ignore_index=True)
```

```python
# 42. df1 və df2-ni sətir indekslərinə görə yan-yana birləşdirin (concat axis=1).
pd.concat(objs=[df1, df2], axis=1)
```

```python
# 43. df1 və df2-ni ortaq 'ID' sütunu üzərindən Inner Join vasitəsilə birləşdirin (merge).
pd.merge(left=df1, right=df2, how="inner", on="ID")
```

```python
# 44. df1 və df2-ni df1-in bütün sətirləri saxlanılmaqla Left Join vasitəsilə birləşdirin.
pd.merge(left=df1, right=df2, how="left", on="ID")
```

```python
# 45. df1 və df2-ni heç bir məlumat itkisi olmadan (Outer Join) birləşdirin.
pd.merge(left=df1, right=df2, how="outer", on="ID")
```

### Bölmə 8: Advanced Texnikalar (Pivot, Apply, String Manipulation)

```python
# 46. Titanic məlumat dəstində sətirlərdə 'Sex', sütunlarda 'Passenger_Class' və içəridə 'Fare'-in orta qiymətini nümayiş etdirən Pivot Cədvəl (pivot_table) yaradın.
df.pivot_table(index="Sex", columns="Passenger_Class", values="Fare")
```

```python
# 47. 'Name' sütunundakı bütün mətnləri kiçik hərflərə (lowercase) çevirin (.str.lower).
df["Name"].str.lower()
```

```python
# 48. Mətn analizi: 'Name' sütununda 'Mr.' titulu keçən sərnişinləri tapın (.str.contains).
mask = df["Name"].str.contains("Mr\.")
df[mask]
```

```python
# 49. `.apply()` metodundan istifadə edərək 'Age' sütunundakı hər bir dəyəri vizual onluqlara çevirin (məsələn, 25 yaşı 2-yə, yəni yaş / 10).
df["Age_Group"] = df["Age"].apply(func=lambda x: str(int(((x // 10) * 10))) + "s")  # I made some changes to the logic of the question :)
df
```

```python
# 50. `lambda` funksiyası ilə mürəkkəb əməliyyat: 'Fare' dəyəri 50-dən böyükdürsə 'Premium', kiçikdirsə 'Standard' yazan yeni bir 'Ticket_Category' sütunu yaradın.
df["Ticket_Category"] = df["Fare"].apply(func=lambda x: "Premium" if x > 50 else "Standard")
df
```
