# Using a Function in Another Function

Define a function that adds 5 to the parameter. Then, define another function that will multiply the newly obtained number by 3. Verify your code was correct by calling the second function with an argument of 5. Was your output equal to 30?

```python
# Method 1: Independent functions
def adder(x: float):
    return x + 5
    
def multiplier(x: float):
    return x*3

multiplier(adder(5))
```

```python
# Method 2: Nested functions
def multiplier(x: float):
    
    def adder(x: float):
        return x + 5
        
    return adder(x) * 3

multiplier(5)
```

```python
def distance_from_zero(x: float):
    """
    Desc:
        This function will calculate the distance of argument from 0 which added by user
    Args:
        x (float): Take user defined argument in order to calculate distance from 0
    Return:
        In case of non-numeric input it will return "Not Possible" string
        In case argument is numeric function performs abs() function in order to find absolute form of numeric argument
    """
    if isinstance(x, float) or isinstance(x, int):
        return abs(x)
    else:
        return "Not Possible"
```

# Iterating over Dictionaries

In this exercise you will use the same dictionaries as the ones we used in the lesson - "prices" and "quantity". This time, don't just calculate all the money Jan spent. Calculate how much she spent on products with a price of 5 dollars or more.

```python
prices = {
    "box_of_spaghetti" : 4,
    "lasagna"  : 5,
    "hamburger" : 2
   }
quantity = {
    "box_of_spaghetti" : 6,
    "lasagna"  : 10,
    "hamburger" : 0
    }

money_spent = 0
```

```python
# 5 dollars or more products
expensive_products = []
expensive_spent = 0

for key in prices.keys():
    money_spent += prices[key] * quantity[key]
    if prices[key] >= 5:
        expensive_products.append(key)
        expensive_spent += prices[key] * quantity[key]
        
print(f"Total Spent: {money_spent}\nExpensive Products: {expensive_products}\nSpent on Expensives: {expensive_spent}")
```

And how much did Jan spent on products that cost less than 5 dollars?

```python
prices = {
    "box_of_spaghetti" : 4,
    "lasagna"  : 5,
    "hamburger" : 2
   }
quantity = {
    "box_of_spaghetti" : 6,
    "lasagna"  : 10,
    "hamburger" : 0
    }

money_spent = 0
```

```python
# Products which cost less than 5 dollars
cheap_products = []
cheap_spent = 0

for key in prices.keys():
    if prices[key] < 5:
        cheap_products.append(key)
        cheap_spent += prices[key] * quantity[key]
        
print(f"Cheap Products: {cheap_products}\nSpent on Cheaps: {cheap_spent}")
```
